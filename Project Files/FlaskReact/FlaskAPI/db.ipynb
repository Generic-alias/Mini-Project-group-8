```python
import pymysql
import pandas as pd
import plotly.express as px
```

```python
db = pymysql.connect(
host = "localhost",
user = "root", #root #aditya
password = "root",
database = "mini"
)
cursor = db.cursor()
```

```python
cursor.execute("select id, sleep_efficiency from sleep_data")
results = cursor.fetchall()
```

```python
results
```

```python
sleep_efficiency_array, id_array = [], []
for i in results:
    sleep_efficiency_array.append(i[1])
    id_array.append(i[0])

sleep_efficiency_array = [i * 100 for i in sleep_efficiency_array]
```

```python
df = pd.DataFrame({"Sleep Efficiency": sleep_efficiency_array,
"ID": id_array})
df
```

```python
fig = px.bar(df, x = "ID", y = "Sleep Efficiency")
fig.update_layout(bargap = 0.3)
fig.show()
```
